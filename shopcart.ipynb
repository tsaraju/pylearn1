```python
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

```python
#Display all products
cart
```

```python
#Access first and last products
print(f'First product: {cart[0]} \nLast product: {cart[-1]}')
```

```python
#Add "Webcam"
cart.append("Webcam")
cart
```

```python
#Insert "USB Hub" at index 2
cart.insert(2, "USB Hub")
cart
```

```python
#Remove "Mouse"
cart.remove("Mouse")
cart
```

```python
#Remove the last item using pop()
cart.pop()
cart
```

```python
#Find the index of "Monitor"
cart.index('Monitor')
```

```python
#Count occurrences of "Laptop"
cart.count('Laptop')
```

```python
#Create a copy of the cart
cart2 = cart.copy()
cart2
```

```python
#Reverse the cart
cart2.reverse()
cart2
```

```python
#Sort the products alphabetically
cart2.sort()
cart2
```

```python
#extend the cart with new products
cart2.extend(["Charger", "Speakers"])
cart2
```

```python
#clear emptying a list
cart2.clear()
cart2
```
