```python
message = '   Welcome to Python Programming class   '
message
```

```python
#Convert to title case
message = message.title()
message
```

```python
#Remove extra spaces
message = message.strip()
message
```

```python
#Convert everything to lowercase
message.lower()
```

```python
#Convert everything to uppercase
message.upper()
```

```python
#Replace "Python" with "Advanced Python"
message.replace("Python","Advanced Python")
```

```python
#Check whether the string starts with "Welcome"
message.startswith("Welcome")
```

```python
#Check whether it ends with "Class"
message.endswith("Class")
```

```python
#Count occurrences of "o"
message.count('o')
```

```python
#Find the position of "Programming"
message.find("Programming")
```

```python
#Split the sentence into words
message.split()
```

Please find the difference between
strip()
lower()
upper()
title()
replace()
find()
count()
split()

strip() is used to remove the spaces at start and end of the string
lower() is used to convert all the letters of the string to lowercase
upper() is used to convert all the letters of the string to uppercase
title() is used to capitalizes the first letter of each word
replace()  is used to replce the given substring with another string
find()  is used to find the position of the first occurence of substring
count()  Counts how many times the given substring appears in the string
split() breaks a string into smaller strings by using delimiter. If no delimiter specified, whitespace takes as default delimiter to break.
