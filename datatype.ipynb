```python
# 15 different variables    
i1 = 10
i2 = 20
f1 = 2.69
f2 = 5020.65
str1 = "Hello"
b1 = True
b2 = False
l1 = [1,2,3,4]
l2 = ['a','b','c','d']
t1 = (5,6,7,8)
t2 = ('e','f','g','h')
s1 = {9,10,11,12}
s2 = {'i','j','k','l'}  
d1 = {"name":"John", "age":30, "city":"New York"}
d2 = {"a":1, "b":2, "c":3, "d":4}
```
