```python
laptop = {
"brand": "Dell",
"model": "XPS 15",
"price": 120000,
"ram": "16GB",
"storage": "512GB SSD",
"available": True
}
```

```python
#Display the product
laptop
```

```python
#Print brand
laptop['brand']
```

```python
#Print model
laptop['model']
```

```python
#Print price
laptop['price']
```

```python
#Change price
laptop['price'] = 130000
laptop
```

```python
#Add processor information
laptop['processor'] = "Intel Core i7"
laptop
```

```python
#Add GPU information
laptop['gpu'] = "NVIDIA GTX 1650"
laptop
```

```python
#Change RAM to "32GB"
laptop['ram'] = "32GB"
laptop
```

```python
#Remove "available"
laptop.pop('available')
laptop
```

```python
#Print all keys
laptop.keys()
```

```python
#Print all values
laptop.values()
```

```python
#Print all items
laptop.items()
```
