```python
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")
```

```python
#Print the tuple
technologies
```

```python
#Print its type
type(technologies)  
```

```python
#Print the first item
technologies[0]
```

```python
#Print the last item
technologies[-1]
```

```python
#Slice the tuple
technologies[2:4]
```

```python
#Count "Python"
technologies.count("Python")
```

```python
#Find the index of "C++"
technologies.index("C++")
```

```python
#Find the tuple length
len(technologies)
```

```python
#Convert the tuple into a list
li1 = list(technologies)
li1
```

```python
#Add "Go" after converting it into a list
li1.append('Go')
li1
```

```python
#Convert it back into a tuple
li1 = tuple(li1)
li1
```

```python
li1.append('Cobol')  # This will raise an AttributeError since tuples are immutable
```

```python
li1[1] = 'Ruby'  # This will also raise an error since tuples do not support item assignment
```
